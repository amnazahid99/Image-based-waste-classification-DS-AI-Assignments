```python
# ============================================================
# 🧠 Week 1 – Environment Setup & Exploration
# ============================================================

# --- 1️⃣  Environment & Library Check
!python --version

import sys, platform
print(f"Python  version : {sys.version}")
print(f"OS / Platform   : {platform.system()} {platform.release()}")

# Core libraries
import tensorflow as tf, keras, cv2, numpy as np, matplotlib.pyplot as plt, os, zipfile, random
from glob import glob
print(f"TensorFlow ver. : {tf.__version__}")
print(f"Keras ver.      : {keras.__version__}")
print(f"OpenCV ver.     : {cv2.__version__}")

# --- 2️⃣  Kaggle Dataset Integration
#  (Make sure kaggle.json is downloaded from your Kaggle account)
os.environ['KAGGLE_CONFIG_DIR'] = r"path/to/your/kaggle.json"

# Download the Garbage Classification v2 dataset (≈400 MB)
!kaggle datasets download -d sumn2u/garbage-classification-v2 -p ./data/

# --- 3️⃣  Unzip Dataset
dataset_zip = "./data/garbage-classification-v2.zip"
extract_path = "./data/garbage_classification_v2"
if os.path.exists(dataset_zip):
    with zipfile.ZipFile(dataset_zip, "r") as z:
        z.extractall(extract_path)
    print("✅ Dataset extracted successfully.")
else:
    print("⚠️ Dataset zip not found – check download step.")

# --- 4️⃣  Folder Structure Verification
for cls in os.listdir(extract_path):
    cls_path = os.path.join(extract_path, cls)
    if os.path.isdir(cls_path):
        print(f"{cls:<12} → {len(os.listdir(cls_path))} images")

# --- 5️⃣  Display Random Sample from Each Class
classes = [c for c in os.listdir(extract_path) if os.path.isdir(os.path.join(extract_path, c))]
plt.figure(figsize=(12,6))
for i, cls in enumerate(classes):
    cls_path = os.path.join(extract_path, cls)
    img_file = random.choice(glob(os.path.join(cls_path, "*.jpg")))
    img = cv2.cvtColor(cv2.imread(img_file), cv2.COLOR_BGR2RGB)
    plt.subplot(2, 3, i+1)
    plt.imshow(img)
    plt.title(cls)
    plt.axis("off")
plt.tight_layout()
plt.show()

# --- 6️⃣  Reflection Summary
print("""
✅ Environment successfully configured.
✅ Kaggle dataset integrated and verified.
✅ All 6 categories detected (Cardboard, Glass, Metal, Paper, Plastic, Trash).
Ready for Week 2: Data Cleaning & Preprocessing.
""")
```
