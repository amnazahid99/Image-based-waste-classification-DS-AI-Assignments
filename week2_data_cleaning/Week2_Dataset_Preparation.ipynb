```python
# Week 2 — Dataset Loading, Organization & Visualization
# Project: Waste Image Classification using CNN

import os
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
import random

# === 1. SETUP PATH ===
DATASET_DIR = r"C:\Users\user\Desktop\jupyter notebook\garbage\garbage-dataset"

print("📂 Dataset exists:", os.path.exists(DATASET_DIR))
classes = sorted(os.listdir(DATASET_DIR))
print("\n🧾 Classes found:", classes)
print("Total classes:", len(classes))

# === 2. COUNT IMAGES PER CLASS ===
class_counts = {}
for cls in classes:
    cls_dir = os.path.join(DATASET_DIR, cls)
    if os.path.isdir(cls_dir):
        count = len([f for f in os.listdir(cls_dir) if f.lower().endswith(('.jpg','.jpeg','.png'))])
        class_counts[cls] = count

print("\n📸 Image Count per Class:")
for k,v in class_counts.items():
    print(f"{k:15} → {v} images")

# === 3. VISUALIZE SAMPLE IMAGES ===
plt.figure(figsize=(12, 8))
for i, cls in enumerate(random.sample(classes, 6)):
    cls_dir = os.path.join(DATASET_DIR, cls)
    sample = random.choice(os.listdir(cls_dir))
    img_path = os.path.join(cls_dir, sample)
    img = Image.open(img_path)
    plt.subplot(2, 3, i + 1)
    plt.imshow(img)
    plt.title(cls)
    plt.axis("off")
plt.suptitle("Random Samples from Waste Dataset")
plt.show()

# === 4. IMAGE SIZE DISTRIBUTION ===
sizes = []
for cls in classes:
    cls_dir = os.path.join(DATASET_DIR, cls)
    for f in os.listdir(cls_dir)[:30]:  # sample 30 per class
        if f.lower().endswith(('.jpg','.jpeg','.png')):
            img_path = os.path.join(cls_dir, f)
            img = Image.open(img_path)
            sizes.append(img.size)

widths, heights = zip(*sizes)
plt.figure(figsize=(10,5))
plt.hist(widths, bins=20, alpha=0.7, label='Width')
plt.hist(heights, bins=20, alpha=0.7, label='Height')
plt.legend()
plt.title("Distribution of Image Dimensions")
plt.xlabel("Pixels")
plt.ylabel("Count")
plt.show()

# === 5. SUMMARY REPORT ===
total_images = sum(class_counts.values())
avg_width = np.mean(widths)
avg_height = np.mean(heights)

summary = f"""
📘 Week 2 — Dataset Summary
---------------------------
Total Classes: {len(classes)}
Total Images: {total_images}

Average Image Size: {avg_width:.1f} x {avg_height:.1f} pixels

Class Distribution:
{class_counts}
"""

print(summary)

# Save summary for GitHub
with open("Week2_Report.txt", "w") as f:
    f.write(summary)

print("\n✅ Saved Week2_Report.txt for GitHub upload.")
```
